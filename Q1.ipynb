```python
import json
import requests
import pandas as pd
```

```python
try:
    with open("api.txt", "r") as f:
        api_url = f.read().strip()
        if not api_url:
            print("Error: Please can you add the url into the txt file")
            exit()
except FileNotFoundError:
    print("Error: file not found, please upload the file")
    exit()
try:
    response = requests.get(api_url, timeout=10)
    if response.status_code != 200:
        print(f"HTTP Error: {response.status_code}")
        print("Response:", response.text)
    else:
        data = response.json()
        print("data fetched successfully")
        print(data) 
except requests.exceptions.RequestException:
    print("Network Error: Please check your internet connection")

if data.get("status") != "success":
    reason = data.get("reason", "Unknown error")
    print(f"API Error: {reason}")
    exit()

matches = data.get("data", [])
if not matches:
    print("No live/current matches found.")
    exit()
```

```python
for match in matches:
    print("="*70)
    print(f"Match Name -> {match.get('name', 'N/A')}")
    print(f"Status     -> {match.get('status', 'N/A')}")
    print(f"Venue      -> {match.get('venue', 'N/A')}")
    print(f"Date       -> {match.get('date', 'N/A')}")
```

```python
scores = match.get("score", [])
if scores:
    print("Scores:")
    for inning in scores:
        print(f"  Inning => {inning.get('inning', 'N/A')}")
        print(f"  Runs   => {inning.get('r', 'N/A')}")
        print(f"  Wickets=> {inning.get('w', 'N/A')}")
        print(f"  Overs  => {inning.get('o', 'N/A')}")
        print("="*70)
    print("----"*15)
```
