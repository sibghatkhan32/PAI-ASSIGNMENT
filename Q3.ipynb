```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
len(df['Age'])
```

```python
Ncol = [i for i in df.columns]
Dcol = []
Nullcol = []
for i in df.columns:
    datatype = str(df[i].dtype)
    count = ((int(df[i].isnull().sum()))/len(df[i]))*100
    Dcol.append(datatype)
    Nullcol.append(count)

# Assuming you have successfully run the loop to generate Ncol, Dcol, and Nullcol

data_dict = {
    'Column Name': Ncol,
    'Data Type': Dcol,
    'Missing (%)': Nullcol
}

inspectreport = pd.DataFrame(data_dict)
with open('inspect_report' , 'w') as f:
    f.write(str(data_dict))
inspectreport
```

```python
df['Age'] = df['Age'].fillna(df.groupby(['Pclass', 'Sex'])['Age'].transform('median'))
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.info()
```

```python
df.drop(labels='Cabin' , axis=1 , inplace=True)
```

```python
df.info()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch']
```

```python
df.head()
```

```python
df['IsAlone'] = (df['FamilySize'] == 0).astype(int)
```

```python
df.head(10)
```

```python
df['Age'] = df['Age'].astype('int64')
```

```python
df.to_csv('titanic_cleaned.csv', index=False, encoding='utf-8')
```
