```python
import pandas as pd
```

```python
df1 = pd.read_csv('titanic_cleaned.csv')
df2 = pd.read_csv('ticket_fares.csv')
```

```python
df1.head()
```

```python
df2.head()
```

```python
Merg_dataframe=df1.merge(df2,left_on='Ticket' , right_on='Ticket')
Merg_dataframe
```

```python
Merg_dataframe['AgeGroup'] =  pd.cut(Merg_dataframe['Age'], bins = [0,12,19,59,200] , labels=['Child', 'Teen', 'Adult' , 'Senior'])
```

```python
Merg_dataframe
```

```python
Merg_dataframe.groupby(['Sex' , 'AgeGroup'])['Survived'].mean()
```

```python
with open('report.txt', 'w') as f:
    f.write('Women show a significantly higher survival rate than men, indicating that gender played a major role in rescue priority.\nChildren also have a noticeably better chance of survival compared to adult men, though their advantage is smaller than that of women.\nWhile not every female or child survived—and some adult men in higher classes did—the numbers clearly lean toward preferential treatment for women and younger passengers during evacuation.')
```

```python
#Test Hypothesis 2
Merg_dataframe.groupby('Pclass')['Survived'].mean()
```

```python
Merg_dataframe['FareBin'] = pd.qcut(Merg_dataframe['Fare_x'] , 4 , labels=['Low', 'Medium' , 'High', 'VeryHigh'])
```

```python
Merg_dataframe
```

```python
Merg_dataframe.groupby('FareBin')['Survived'].mean()
```
