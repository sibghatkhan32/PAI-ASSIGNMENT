```python
import numpy as np
```

```python
fi = []
with open('sensor_data.csv', 'r') as f:
    for i in f:
        data = i.strip().split(',')
        fi.append([float(x) for x in data])
```

```python
n_array = np.array(fi)
```

```python
n_array
```

```python
n_array[n_array == -999] = np.nan
```

```python
n_array[(n_array < 0) & (n_array > 100)] = np.nan
```

```python
n_array
```

```python
ms = np.nanmean(n_array,axis=0) # for columns we use axis = 0
```

```python
print("mean moisture for each sensor is: ",ms)
```

```python
mes = np.nanmedian(n_array, axis = 1)
```

```python
print("the meidan moisture for each hour is: ",mes)
```

```python
nan_count=np.sum(np.isnan(n_array) , axis = 0)
nan_count.shape
```

```python
np.argmax(nan_count)
```

```python
m = np.nanmax(n_array)
n = np.nanmin(n_array)
print("maximum reading is: ",m,"\n","minimum reading is: ",n)
```

```python
x_norm = (n_array - n)/(m - n)
```

```python
x_norm[:5]
```

```python
np.savetxt('sensor_data_normalized.csv' ,x_norm, delimiter=',', fmt='%.4f')
```
